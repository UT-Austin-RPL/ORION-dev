```python
import math

tmp = mesh_other.copy()
angle = -math.pi / 2
direction = [1, 0, 0]
center = tmp.centroid

rot_matrix = trimesh.transformations.rotation_matrix(angle, direction, center)
tmp.apply_transform(rot_matrix)

rot_matrix = trimesh.transformations.rotation_matrix(-math.pi/2, [0, 1, 0], center)
tmp.apply_transform(rot_matrix)

tmp.show(flags={'wireframe': True, 'axis': True})
```

```python
to_origin_instantmesh, extents_instantmesh = trimesh.bounds.oriented_bounds(mesh_other, ordered=False)
to_origin_colmap, extents_colmap = trimesh.bounds.oriented_bounds(mesh_colmap, ordered=False)
print(extents_colmap, extents_instantmesh)
```

```python
#scale other mesh so that the extents are the same as colmap

# Get the axis-aligned bounding box (AABB) for mesh A
aabb_colmap = mesh_colmap.bounds
length_A_x = aabb_colmap[1][0] - aabb_colmap[0][0]
length_A_y = aabb_colmap[1][1] - aabb_colmap[0][1]
length_A_z = aabb_colmap[1][2] - aabb_colmap[0][2]
lengths_colmap = [length_A_x, length_A_y, length_A_z]

# Get the axis-aligned bounding box (AABB) for mesh B
aabb_other = mesh_other.bounds
length_B_x = aabb_other[1][0] - aabb_other[0][0]
length_B_y = aabb_other[1][1] - aabb_other[0][1]
length_B_z = aabb_other[1][2] - aabb_other[0][2]
lengths_other = [length_B_x, length_B_y, length_B_z]
scale = np.divide(lengths_colmap, lengths_other)
print(lengths_colmap, lengths_other)
print(scale)

mesh_other.apply_scale(scale)
mesh_other.show()
```
